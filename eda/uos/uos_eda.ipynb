## 1. Load the data

```python
from pathlib import Path

from harness.data import canonical


def repo_root():
    """The repository root: the first folder upwards that contains pyproject.toml."""
    p = Path.cwd().resolve()
    while not (p / "pyproject.toml").exists():
        p = p.parent
    return p


ROOT = repo_root()
DATA_ROOT = ROOT / "data" / "canonical"  # as data_root in configs/datasets/
DATASET = "uos"

# The four canonical tables (runs, captures, channels, telemetry) for all converted runs.
meta = canonical.load_metadata(DATA_ROOT, DATASET)
for name, table in meta.items():
    print(f"{name}: {len(table)} rows")

# One signal: the first channel of the first capture, in physical units.
first = meta["captures"].iloc[0]
channel = meta["channels"].query("run_id == @first.run_id").channel.iloc[0]
x = canonical.load_signal(DATA_ROOT, DATASET, first.run_id, first.capture_id, channel)
print(first.run_id, first.capture_id, channel, x.shape)
```
